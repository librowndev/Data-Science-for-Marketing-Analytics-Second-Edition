### Activity 5.02 Predicting Store Revenue Using Linear Regression

The goal of this activity is build a predictive model using linear regress to predict the revenue of a store using information about its locations. Using the train-test-split approach to train the model on train data and assess the performance on unseen test data. Calculate the correlation between actual and predicted values. Additionally, examine the coefficients of the model to determine which variables have the most influence on revenue.

```python
location_rev.head()
```

```python
# assgn predictior columns to variable X (all columns but revenue) and assign the target column to variable y (revenue)
X = location_rev.drop('revenue', axis=1)

y = location_rev['revenue']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train, y_train)

print(f"model coefficients are:  {lr.coef_} ")
print(f"model intercept is  {lr.intercept_}")
```

```python
# predict correlation between predictual and actual revenue_2020 values
np.corrcoef(lr.predict(X_test), y_test)
```

```python
# plot the predicted values against the actual values
sns.regplot(x=lr.predict(X_test), y=y_test, line_kws={"color": "red"})
plt.xlabel('Predicted Revenue')
plt.ylabel('Actual Revenue')
plt.title('Predicted vs Actual Revenue')
plt.show()
```

```python
# plotly chart

fig = px.scatter(x=lr.predict(X_test), y=y_test, trendline='ols', trendline_color_override="red")
fig.update_layout(title='Predicted vs Actual Revenue',
                  xaxis_title='Predicted Revenue',
                  yaxis_title='Actual Revenue')

fig.show(renderer="notebook")
```
