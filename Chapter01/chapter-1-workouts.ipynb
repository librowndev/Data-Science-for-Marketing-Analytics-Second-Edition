### Complete the Activity

```python
sales =  pd.read_csv(r"Activity1.01/sales.csv", sep = ",")
sales.head()
```

```python
sales['order_method']= sales['type.1']
sales['product_line'] = sales['Product'] + ' ' + sales['line']
sales['product_type'] = sales['Product.1'] + ' ' + sales['type']
sales['Product1'] = sales['Product.2'] + ' ' + sales['Order'] + ' ' + sales['method']
sales['country'] = sales['country'].fillna(' ')
sales['retailer_country']= sales['Retailer'] + ' ' + sales['country']
sales = sales.drop(columns=['Product', 'line', 'Product.1', 'type', 'Product.2', 'Order', 'method', 'type.1', 'Retailer', 'country'], axis=1)

sales.head()
```

```python
sales.describe()
```
